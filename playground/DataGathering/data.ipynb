```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

# URL of the page
url = "https://idsp.mohfw.gov.in/index4.php?lang=1&level=0&linkid=406&lid=3689"

# Headers to mimic a browser request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

# Send a GET request
response = requests.get(url, headers=headers)

# Check if request was successful
if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    
    data = []
    current_year = None  # Track the year

    # Iterate through all elements
    for element in soup.find_all(["strong", "a"]):
        if element.name == "strong":  # Year is usually in <strong> tags
            current_year = element.text.strip()
        elif element.name == "a" and "href" in element.attrs:  # PDF links in <a> tags
            week = element.text.strip()
            link = element["href"]
            if "drive.google.com" in link or "WriteReadData" in link:  # Filtering relevant links
                data.append({"Year": current_year, "Week": week, "Link": link})

    # Save to CSV
    df = pd.DataFrame(data)
    df.to_csv("pdf_links_with_years.csv", index=False)

    print(f"\nExtracted {len(data)} links and saved to pdf_links_with_years.csv")
else:
    print("Failed to retrieve the webpage")
```
