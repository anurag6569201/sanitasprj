```python
import os
import pandas as pd

download_folder = "downloads_2024"
os.makedirs(download_folder, exist_ok=True)

def get_file_id(url):
    """Extract Google Drive file ID from URL."""
    if "id=" in url:
        return url.split("id=")[-1]
    elif "/d/" in url:
        return url.split("/d/")[1].split("/")[0]
    return None

def download_drive_file(file_id, filename):
    """Download Google Drive file using wget."""
    file_path = os.path.join(download_folder, filename)
    download_url = f"https://drive.google.com/uc?export=download&id={file_id}"
    
    command = f"wget --no-check-certificate '{download_url}' -O '{file_path}'"
    os.system(command)
    print(f"✅ Downloaded: {file_path}")

# Download all files
for index, row in df_2024.iterrows():
    file_id = get_file_id(row["Link"])
    if file_id:
        filename = f"Week_{row['Week']}_2024.pdf"
        download_drive_file(file_id, filename)
    else:
        print(f"⚠️ Invalid Google Drive link: {row['Link']}")

print(f"📁 All files are saved in: {download_folder}")
```
